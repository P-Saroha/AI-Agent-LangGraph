```python
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from pydantic import BaseModel, Field
```

```python
load_dotenv(find_dotenv())
```

```python
llm = ChatGoogleGenerativeAI(
            model="gemini-2.5-flash",
            temperature=0.3
              )
```

```python
class EvaluationSchema(BaseModel):

    feedback: str = Field(description= "Give Detailed feedback for the given eassy")
    score: int = Field(description= 'Score out of 10', ge=0, le=10)
```

```python
structured_model =  llm.with_structured_output(EvaluationSchema)
```

```python
essay = """"
        Can AI Outsmart Humans in the Future
Introduction
Artificial Intelligence (AI) has emerged as one of the most transformative technologies of the 21st century. From healthcare diagnostics to autonomous vehicles, AI systems are increasingly performing tasks that were once considered the exclusive domain of human intelligence. This raises a fundamental question: Can AI eventually surpass human intelligence and decision-making capabilities? The answer requires a nuanced exploration of technological progress, ethical considerations, and the essence of human cognition.

The Rise of Artificial Intelligence
- Automation of routine tasks: AI already outperforms humans in repetitive, data-driven activities such as pattern recognition, fraud detection, and predictive analytics.
- Advances in machine learning: With deep learning and neural networks, AI systems can process vast datasets far beyond human capacity.
- Generative AI: Tools capable of creating text, images, and even music demonstrate creativity-like outputs, challenging traditional notions of human uniqueness.

Areas Where AI May Outsmart Humans
- Speed and Scale: AI can analyze millions of variables in seconds, a feat impossible for humans.
- Accuracy in specialized domains: In medical imaging or weather forecasting, AI often achieves higher precision than human experts.
- Memory and consistency: Unlike humans, AI does not suffer from fatigue, bias, or forgetfulness.

Limitations of AI
- Lack of consciousness: AI operates on algorithms and data; it does not possess emotions, intuition, or moral reasoning.
- Dependence on data quality: Biased or incomplete datasets can lead to flawed outcomes.
- Ethical dilemmas: Decisions in law, governance, or warfare cannot be left solely to machines.

Human Strengths That AI Cannot Replicate
- Creativity and imagination: True innovation often arises from intuition, serendipity, and emotional depth.
- Ethical judgment: Humans weigh values, empathy, and cultural context in decision-making.
- Adaptability: Unlike AI, humans can thrive in uncertain, novel situations without prior data.

Future Possibilities
- Collaborative intelligence: The most likely future is not AI replacing humans, but AI augmenting human capabilities.
- Superintelligence debate: Some theorists argue that AI could evolve into a form of “superintelligence” surpassing human cognition, while others believe human consciousness is irreplaceable.
- Policy and governance: The trajectory of AI will depend on how societies regulate its development, ensuring it remains a tool for empowerment rather than domination.

Conclusion
AI may outsmart humans in specific domains of speed, accuracy, and scale, but it cannot replicate the holistic essence of human intelligence—creativity, morality, and emotional depth. The future lies not in competition but in synergy, where AI serves as a powerful ally to human progress. The challenge for policymakers, technologists, and citizens is to harness AI responsibly, ensuring that it complements rather than threatens humanity.
"""
```

```python
prompt = f'You are UPSC Essay Evaluater provide an feedback also give score beteween 0-10 \n {essay}'
structured_model.invoke(prompt)
```
